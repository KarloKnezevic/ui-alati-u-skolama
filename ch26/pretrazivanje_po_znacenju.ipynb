# Pretraživanje po značenju nad opisima listića

Poglavlje 26. U lijevom stupcu otvorite ploču s datotekama i prenesite `listici_izmisljeni.txt` iz spremišta ili svoju datoteku (tada u `DATOTEKA` upišite njezin naziv). Prvo pokretanje traje nekoliko minuta jer se preuzima model. Ništa o učenicima: Colab je Googleov oblak.


```python
!pip install -q sentence-transformers
from sentence_transformers import SentenceTransformer

# POSTAVKE: za usporedbu upišite "BAAI/bge-m3"
MODEL = "Qwen/Qwen3-Embedding-0.6B"
DATOTEKA = "listici_izmisljeni.txt"   # vlastita: npr. "listici.txt"
retci = open(DATOTEKA, encoding="utf-8").read().splitlines()
listici = [r for r in retci if r.strip()]

model = SentenceTransformer(MODEL)
vektori = model.encode_document(listici)   # svaki redak: 1024 broja

def trazi(upit, koliko=5):
    v = model.encode_query(upit)
    slicnost = model.similarity(v, vektori)[0]   # sa svim retcima
    for i in slicnost.argsort(descending=True)[:koliko]:
        print(f"{slicnost[i]:.2f}  {listici[i].split('|')[0]}")

trazi("razlomci")
```

```python
# Pet pitanja na koja znate odgovor; zapišite je li očekivani listić u prvih pet.
trazi("rješavanje jednadžbi")
trazi("kako izmjeriti kut")
```
